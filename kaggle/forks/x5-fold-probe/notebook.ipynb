```python
import glob, json, os
def find(p): return sorted(glob.glob('/kaggle/input/**/' + p, recursive=True))
d = os.path.dirname(find('fpnet_fold0.pt')[0])
print('DIR', d, sorted(os.listdir(d)))
for name in ('README.md', 'SPLIT_AUDIT.json', 'MODEL_INPUTS.json', 'SOURCE_CHECK_20261019.md'):
    print('=====', name); print(open(os.path.join(d, name)).read()[:7000])
for name in ('check_inputs_fold0.json', 'FIVEFOLD_VALIDATION.json', 'PANEL_REPLAY_PUBLIC.json', 'MANIFEST_FIVEFOLD.json'):
    obj = json.load(open(os.path.join(d, name)))
    def shape(o, depth=0):
        if isinstance(o, dict):
            return {k: shape(v, depth + 1) for k, v in list(o.items())[:12]} if depth < 3 else f'dict[{len(o)}]'
        if isinstance(o, list):
            return [f'list[{len(o)}]', shape(o[0], depth + 1) if o else None]
        return repr(o)[:80]
    print('=====', name); print(json.dumps(shape(obj), indent=1)[:4000])
for p in find('catalogue_experiment/*') + find('catalogue_replay_checkpoint/*')[:20]:
    print('SANER', p)
```
